```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv",skiprows=4)
print(df.head())
```

```python
year_cols = [col for col in df.columns if col.isdigit()]

#from last country
latest_year = year_cols[-1]

# remove null data 

data = df[['Country Name', latest_year]].dropna()

data[latest_year] = pd.to_numeric(data[latest_year])

top10 = data.sort_values(by=latest_year, ascending=False).head(10)
```

```python
plt.figure()
plt.bar(top10['Country Name'],top10[latest_year])
plt.xlabel("Country")
plt.ylabel('population')
plt.title('Top 10 countary population ({latest_year})')
plt.xticks(rotation =45)
plt.show()
```

Histogram

```python

population = pd.to_numeric(df[latest_year], errors='coerce').dropna()

# Plot histogram
plt.figure()
plt.hist(population, bins=20)
plt.xlabel("Population")
plt.ylabel("Number of Countries")
plt.title(f"Population Distribution ({latest_year})")
plt.show()
```
